```python
import sys
from pathlib import Path
import pickle
import numpy as np

PROJECT_ROOT = Path.cwd().parents[1]
sys.path.append(str(PROJECT_ROOT))

from src.second_encoding import SecondEncoder
from src.quantum.state_construction import StateConstructionBuilder
```

```python
with open(PROJECT_ROOT / "outputs" / "project_data.pkl", "rb") as f:
    project_data = pickle.load(f)
```

```python
with open(
    PROJECT_ROOT /
    "outputs" /
    "candidate_evaluation" /
    "matched_candidates.pkl",
    "rb",
) as f:
    matched_candidates = pickle.load(f)
```

```python
encoder = SecondEncoder(
    project_data.interaction_ranges
)
```

```python
ligand_interactions = tuple(

    (
        row["Hydrophobic"],
        row["Hydrogen_Bond"]
    )

    for _, row in project_data.ligand.iterrows()

)

ligand_interactions
```

```python
candidate = matched_candidates[0]

encoding_result = encoder.build(
    candidate,
    ligand_interactions,
)

encoding_result
```

```python
print("Occurrence :", encoding_result.occurrence)
print("Probability:", encoding_result.probability)

print()

print("Ligand Dimension :", len(encoding_result.ligand_vector))
print("Protein Dimension:", len(encoding_result.protein_vector))
```

```python
builder = StateConstructionBuilder()

state_result = builder.build(
    encoding_result
)
```

```python
print("Norm(A)")
print(np.linalg.norm(
    encoding_result.ligand_vector
))

print()

print("Norm(B)")
print(np.linalg.norm(
    encoding_result.protein_vector
))
```
