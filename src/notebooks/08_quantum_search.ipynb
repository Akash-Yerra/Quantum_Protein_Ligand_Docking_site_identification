```python
import os
import sys
import pickle

project_root = os.path.abspath(
    os.path.join(os.getcwd(), "..", "..")
)

if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.quantum.quantum_search import QuantumSearch
```

```python
with open("../../outputs/project_data.pkl", "rb") as f:
    project_data = pickle.load(f)
```

```python
def ligand_to_bitstring(encoded_ligand):
    return "".join(f"{a}{b}" for a, b in encoded_ligand["Encoded"])

ligand_bits = ligand_to_bitstring(project_data.encoded_ligand)

print(ligand_bits)
```

```python
search = QuantumSearch()
output = search.search(
    shift_id=1,
    amplitude_data=project_data.amplitude_vectors[1],
    target_state=ligand_bits,
    backend=None,
    hardware="simulator",
    iterations=1,
    shots=1024,
)
```

```python
print("Shift :", output.shift_id)
print("Threshold :", output.threshold)
print("Counts :", output.counts)
```

```python
for result in output.results:
    print("="*50)
    print("Pattern      :", result.pattern.binary_window)
    print("Occurrences  :", result.pattern.occurrences)
    print("Counts       :", result.counts)
    print("Probability  :", result.probability)
```

```python
assert sum(output.counts.values()) == 1024
```

```python
prob = sum(
    result.probability
    for result in output.results
)

print(prob)
```

```python
assert len(output.results) == len(project_data.amplitude_vectors[1].patterns)
```
